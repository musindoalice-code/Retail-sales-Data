```sql
SELECT 
  `Transaction ID`,
  CAST(Date AS DATE) AS date,
  `Customer ID`,
  Gender,
  Age,
  CASE 
    WHEN Age < 25 THEN 'Under 25'
    WHEN Age BETWEEN 25 AND 34 THEN '25-34'
    WHEN Age BETWEEN 35 AND 44 THEN '35-44'
    WHEN Age BETWEEN 45 AND 54 THEN '45-54'
    ELSE '55+'
  END AS age_group,
  `Product Category`,
  Quantity,
  `Price per Unit`,
  `Total Amount`,
  DATE_TRUNC('month', CAST(Date AS DATE)) AS month
FROM sales.default.sales_retail_dataset
--KPI Summary (For Counter Widgets)
SELECT 
  SUM(`Total Amount`) AS total_revenue,
  COUNT(*) AS total_transactions,
  COUNT(DISTINCT `Customer ID`) AS unique_customers,
  AVG(`Total Amount`) AS avg_transaction_value,
  SUM(Quantity) AS total_quantity_sold,
  AVG(Quantity) AS avg_items_per_transaction
FROM sales.default.sales_retail_dataset
-- Revenue by Product Category (Bar Chart)
SELECT 
  `Product Category` AS product_category,
  SUM(`Total Amount`) AS total_revenue,
  COUNT(*) AS total_transactions,
  SUM(Quantity) AS total_quantity_sold
FROM sales.default.sales_retail_dataset
GROUP BY `Product Category`
ORDER BY total_revenue DESC
--Revenue Trend Over Time (Line Chart)
SELECT 
  DATE_TRUNC('month', CAST(Date AS DATE)) AS month,
  SUM(`Total Amount`) AS total_revenue,
  COUNT(*) AS total_transactions
FROM sales.default.sales_retail_dataset
GROUP BY DATE_TRUNC('month', CAST(Date AS DATE))
ORDER BY month
--Sales by Gender (Pie Chart)
SELECT 
  Gender,
  SUM(`Total Amount`) AS total_revenue,
  COUNT(*) AS total_transactions,
  COUNT(DISTINCT `Customer ID`) AS unique_customers
FROM sales.default.sales_retail_dataset
GROUP BY Gender
--Revenue by Age Group (Bar Chart)
SELECT 
  CASE 
    WHEN Age < 25 THEN 'Under 25'
    WHEN Age BETWEEN 25 AND 34 THEN '25-34'
    WHEN Age BETWEEN 35 AND 44 THEN '35-44'
    WHEN Age BETWEEN 45 AND 54 THEN '45-54'
    ELSE '55+'
  END AS age_group,
  SUM(`Total Amount`) AS total_revenue,
  COUNT(*) AS total_transactions
FROM sales.default.sales_retail_dataset
GROUP BY 
  CASE 
    WHEN Age < 25 THEN 'Under 25'
    WHEN Age BETWEEN 25 AND 34 THEN '25-34'
    WHEN Age BETWEEN 35 AND 44 THEN '35-44'
    WHEN Age BETWEEN 45 AND 54 THEN '45-54'
    ELSE '55+'
  END
ORDER BY 
  CASE age_group
    WHEN 'Under 25' THEN 1
    WHEN '25-34' THEN 2
    WHEN '35-44' THEN 3
    WHEN '45-44' THEN 4
    ELSE 5
  END
--Product Category Performance Table
SELECT 
  `Product Category` AS product_category,
  SUM(`Total Amount`) AS total_revenue,
  COUNT(*) AS total_transactions,
  SUM(Quantity) AS total_quantity_sold,
  AVG(`Total Amount`) AS avg_transaction_value,
  COUNT(DISTINCT `Customer ID`) AS unique_customers
FROM sales.default.sales_retail_dataset
GROUP BY `Product Category`
ORDER BY total_revenue DESC
--Customer Repeat Purchase Analysis
SELECT 
  `Customer ID`,
  COUNT(*) AS purchase_count,
  SUM(`Total Amount`) AS customer_lifetime_value,
  MIN(CAST(Date AS DATE)) AS first_purchase_date,
  MAX(CAST(Date AS DATE)) AS last_purchase_date
FROM sales.default.sales_retail_dataset
GROUP BY `Customer ID`
ORDER BY customer_lifetime_value DESC



SELECT * FROM `sales1`.`default`.`retail_sales_dataset_1` limit 100;


SELECT*
FROM sales1.default.retail_sales_dataset_1
LIMIT 100;

SELECT 
  MIN(`Age`) AS min_age,
  MAX(`Age`) AS max_age,
  AVG(`Age`) AS mean_age
FROM sales1.default.retail_sales_dataset_1

SELECT COUNT(*) AS number_of_rows,
FROM sales1.default.retail_sales_dataset_1

SELECT DISTINCT 
CASE 
  WHEN `Age` = '0' THEN 'Infant'
  WHEN `Age`BETWEEN 1 AND 12 THEN 'Kids'
  WHEN `Age`BETWEEN 13 AND 17 THEN 'Youth'
  WHEN `Age`BETWEEN 18 AND 35 THEN 'Young Adult'
  WHEN `Age`BETWEEN 36 AND 50 THEN 'Adults'
  WHEN `Age`>50 AND `Age`<=60 THEN 'Elder'
  WHEN `Age`>64 THEN 'Pensioner'
END AS ethnicity
FROM sales1.default.retail_sales_dataset_1;



```
